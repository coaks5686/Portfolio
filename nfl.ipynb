```python
# Download the NFL data using nflreadpy and visualize it using pandas and matplotlib
import nflreadpy as nfl
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Viewing all the data and columns available in the dataset
pbp = nfl.load_pbp(seasons=[2025])
pbp.head(10)
```

```python
# Transforming the data into a pandas DataFrame for easier manipulation and analysis
df = pbp.to_pandas()
df.head(5)
```

```python
# View Column Names
print(df.columns.tolist())
```

```python
# There are a lot of columns in the dataset and this looks for keywords
for column in df.columns:
    if "yards" in column.lower():
        print(column)
```

```python
# We group the data by rusher player name and sum the rushing yards, then sort the values in descending order and display the top 25 rushers
topRushers = df.groupby('rusher_player_name')['rushing_yards'].sum().sort_values(ascending=False).head(25)
print(topRushers)
```

```python
plt.figure(figsize=(12, 6))
topRushers.plot(kind= 'bar', color='red', rot=0)
plt.title('Top 25 Rushers in the 2025 Season')
plt.xlabel('Player Name')
plt.ylabel('Total Rushing Yards')
plt.xticks(rotation=45)
plt.yticks(np.arange(0, topRushers.max() + 100, 100))

plt.show()
```

```python
topRecievers = df.groupby('receiver_player_name')['receiving_yards'].sum().sort_values(ascending = False).head(25)
print(topRecievers)
```

```python
plt.figure(figsize=(12, 6))
plt.title('Top 25 Receiving Leaders in the 2025 Season')
topRecievers.plot(kind = 'bar', color = 'green', rot = 0)
plt.xlabel('Player Name')
plt.ylabel('Total Receiving Yards')
plt.xticks(rotation = 45)
plt.yticks(np.arange(0, topRecievers.max() + 100, 100))

plt.show()
```

```python
rushrecyards = df.groupby('posteam')[['receiving_yards', 'rushing_yards']].sum()
print(rushrecyards)
```

```python
totalyards=rushrecyards['receiving_yards'] + rushrecyards['rushing_yards'].sort_values(ascending=False)
print(totalyards)
```

```python
plt.figure(figsize=(18,12))
totalyards.plot(kind='pie', autopct='%1.1f%%', startangle=90, shadow=True)
plt.title('Total Yards by Team')
plt.show()
```
