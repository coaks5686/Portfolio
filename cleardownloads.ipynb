```python
from pathlib import Path
import shutil
```

```python
downloads = Path.home() / "Downloads"
print(downloads)
```

```python
for item in downloads.iterdir():
    print(item)
```

```python
for item in downloads.iterdir():
    if "Pokemon" in item.name:
        continue
    if item.is_file():
        print(item.name, "->", item.suffix)
    elif item.suffix == "pptx":
        print(item.name, "->", item.suffix)
    elif item.suffix == "pdf":
        print(item.name, "->", item.suffix)
    elif item.suffix == "jpg":
        print(item.name, "->", item.suffix)
    elif item.suffix == "png":
        print(item.name, "->", item.suffix)
    elif item.suffix == "exe":
        print(item.name, "->", item.suffix)
    elif item.suffix == "zip":
        print(item.name, "->", item.suffix)
    elif item.suffix == "save":
        print(item.name, "->", item.suffix)
```

```python
documents = downloads / "Documents"
powerpoint = downloads / "PowerPoint"
save = downloads / "Save"
installers = downloads / "Installers"
images = downloads / "Images"
resume = downloads / "Resume"
other = downloads / "Other"

documents.mkdir(exist_ok=True)
powerpoint.mkdir(exist_ok=True)
save.mkdir(exist_ok=True)
installers.mkdir(exist_ok=True)
images.mkdir(exist_ok=True)
resume.mkdir(exist_ok=True)
other.mkdir(exist_ok=True)

for item in downloads.iterdir():
    if "pokemon" in item.name.lower():
        continue
    
    if item.is_file():
        if item.suffix == ".pptx":
            shutil.move(item, powerpoint / item.name)
        elif item.suffix == ".txt":
            continue
        elif item.suffix == ".pdf":
            shutil.move(item, documents / item.name)
        elif item.suffix == ".jpg" or item.suffix == ".png":
            shutil.move(item, images / item.name)
        elif item.suffix == ".save":
            shutil.move(item, save / item.name)
        elif item.suffix == ".exe" or item.suffix == ".zip":
            shutil.move(item, installers / item.name)
        elif "resume" in item.name.lower():
            shutil.move(item, resume / item.name)
        else:
            shutil.move(item, other / item.name)
```

```python
for item in other.iterdir():
    if item.is_file():
        if item.suffix == ".xlsx" or item.suffix == ".docx":
            shutil.move(item, documents / item.name)
```
